# Rescaling Atomic units

### Notation: 
one can say m=5kg than {m} = 5 and [m] = kg in other words { } means value and [ ] unit.

HAU = Hartree Atomic Units

RAU = Rescaled Atomic Units

### Atomic base units:
- action: $\hbar = 1.054571817 \times 10−34 Js$ 

\begin{align*}
\{\hbar\}_{SI} &= 1.054571817 \times 10^{-34}; \quad [\hbar]_{SI} = Js \\
\{\hbar\}_{HAU} &= 1; \quad [\hbar]_{HAU} = 1.054571817 \times 10^{-34} Js\\
\{\hbar\}_{RAU} &= X_h; \quad [\hbar]_{RAU} = \underbrace{\frac{1}{X_h}}_{r_h} 1.054571817 \times 10^{-34} Js
\end{align*}

- elementary charge: $e_0 = 1.602176634 \times 10^{−19} C$

\begin{align*}
\{e_0\}_{SI} &= 1.602176634 \times 10^{-19}; \quad [e_0]_{SI} = C \\
\{e_0\}_{HAU} &= 1; \quad [e_0]_{HAU} = 1.602176634 \times 10^{-19} C\\
\{e_0\}_{RAU} &= X_e; \quad [e_0]_{RAU} = \underbrace{\frac{1}{X_e}}_{r_e} 1.602176634 \times 10^{-19} C 
\end{align*}

- electron rest mass: $m_e = 9.1093837139(28) \times 10^{−31} kg$

\begin{align*}
\{m_e\}_{SI} &= 9.1093837139 \times 10^{-31}; \quad [m_e]_{SI} = kg \\
\{m_e\}_{HAU} &= 1; \quad [m_e]_{HAU} = 9.1093837139 \times 10^{-31} kg\\
\{m_e\}_{RAU} &= X_m; \quad [m_e]_{RAU} = \underbrace{\frac{1}{X_m}}_{r_m} 9.1093837139 \times 10^{-31} kg
\end{align*}

- permittivity: $\kappa_0 = 4 \pi \epsilon_0 = 1.11265005620 \times 10^{−10} F/m$

\begin{align*}
\{\kappa_0\}_{SI} &= 1.11265005620 \times 10^{-10}; \quad [\kappa_0]_{SI} = F/m \\
\{\kappa_0\}_{HAU} &= 1; \quad [\kappa_0]_{HAU} = 1.11265005620 \times 10^{-10} F/m\\
\{\kappa_0\}_{RAU} &= X_k; \quad [\kappa_0]_{RAU} = \underbrace{\frac{1}{X_k}}_{r_k} 1.11265005620 \times 10^{-10} F/m
\end{align*}

### Derived units

- length:
  Bohr radius
  \begin{align*} ⁠ 
  a_{0} = \frac{\kappa_0 \hbar ^{2}}{m_{e}e_0^{2}} &= \frac{\{\kappa_0\}_{SI} [\kappa_0]_{SI} \, (\{\hbar\}_{SI} [\hbar]_{SI})^2 }{ \{m_e\}_{SI} [m_e]_{SI} \, (\{e_0\}_{SI} [e_0]_{SI})^{2}}\\
  &= \frac{\{\kappa_0\}_{SI} \, \{\hbar\}_{SI}^2 }{ \{m_e\}_{SI} \, \{e_0\}_{SI}^{2}} \, \frac{ [\kappa_0]_{SI} \, [\hbar]_{SI}^2 }{ [m_e]_{SI} \, [e_0]_{SI}^{2}}\\
  &= \frac{1.11265005620 \times 10−10 \cdot (1.054571817 \times 10−34)^2 }{ 9.1093837139 \times 10−31 \cdot (1.602176634 \times 10−19)^{2}} \, \frac{ F/m \, (Js)^2 }{ kg \, C^2}\\
  &= 5.2917720989499097 \times 10^{-11} m\\
  &\Rightarrow \{ a_{0} \}_{SI}=5.2917720989499097 \times 10^{-11}; \qquad  [a_{0}]_{SI}=m\\
  \end{align*}

  \begin{align*}
  a_{0} &= \frac{\{\kappa_0\}_{HAU} \, \{\hbar\}_{HAU}^2 }{ \{m_e\}_{HAU} \, \{e_0\}_{HAU}^{2}} \, \frac{ [\kappa_0]_{HAU} \, [\hbar]_{HAU}^2 }{ [m_e]_{HAU} \, [e_0]_{HAU}^2}\\
  &= \frac{ 1 \cdot 1^2 }{ 1 \cdot 1^2 } \, \frac{1.11265005620 \times 10^{-10} F/m \, (1.054571817 \times 10−34 Js)^2 }{ 9.1093837139 \times 10^{−31} kg \,(1.602176634 \times 10^{-19} C)^2}\\
  &= 1 a_0 = 1 \, \textup{l}_{HAU}\\
  &\Rightarrow \{ a_{0} \}_{HAU}=1; \qquad  [a_{0}]_{HAU}=5.2917720989499097 \times 10^{-11} m = a_0
  \end{align*}

  \begin{align*}
  a_{0} &= \frac{\{\kappa_0\}_{RAU} \, \{\hbar\}_{RAU}^2 }{ \{m_e\}_{RAU} \, \{e_0\}_{RAU}^{2}} \, \frac{ [\kappa_0]_{RAU} \, [\hbar]_{RAU}^2 }{ [m_e]_{RAU} \, [e_0]_{RAU}^2}\\
  &= \frac{ X_k \cdot X_h^2 }{ X_m \cdot X_e^2 } \, \frac{r_e 1.11265005620 \times 10^{-10} F/m \, (r_h 1.054571817 \times 10−34 Js)^2 }{ r_m 9.1093837139 \times 10^{−31} kg \,(r_e 1.602176634 \times 10^{-19} C)^2}\\
  &= \frac{ X_k \cdot X_h^2 }{ X_m \cdot X_e^2 } \, \frac{r_e \, r_h^2 }{ r_m \,r_e^2} * a_0\\
  &= \frac{ X_k \cdot X_h^2 }{ X_m \cdot X_e^2 } \, \textup{l}_{RAU}\\
  &\Rightarrow \{ a_{0} \}_{RAU} = \frac{ X_k \cdot X_h^2 }{ X_m \cdot X_e^2 } = X_a ; \qquad  [a_{0}]_{RAU}=\frac{r_k \, r_h^2 }{ r_m \,r_e^2} a_0 = r_a \, a_0
  \end{align*}

### Derived units

- Energy:
  Hartree energy $E_h$

\begin{align*}
E_h &= \frac{\hbar^2}{m_e a_0^2} \\
    &= \frac{\{\hbar\}_{SI}^2}{\{m_e\}_{SI} \{a_0\}_{SI}^2} \; \frac{[\hbar]_{SI}^2}{ [m_e]_{SI} [a_0]_{SI}^2} \\
    &= \frac{(1.054571817 \times 10^{-34})^2}{(9.1093837139 \times 10^{-31}) (5.2917720989499097 \times 10^{-11})^2} \, \frac{(\text{Js})^2}{\text{kg} \, \text{m}^2} \\
    &= 4.3597447222071 \times 10^{-18} \, \text{J} \\
    &\Rightarrow \{E_h\}_{SI} = 4.3597447222071 \times 10^{-18}; \qquad [E_h]_{SI} = \text{J}
\end{align*}

\begin{align*}
E_h &= \frac{\{\hbar\}_{HAU}^2}{\{m_e\}_{HAU} \{a_0\}_{HAU}^2} \; \frac{ [\hbar]_{HAU}^2}{ [m_e]_{HAU} [a_0]_{HAU}^2} \\
    &= \frac{1^2 \cdot 1^2}{1 \cdot 1 \cdot 1^2 \cdot 1^2} \, \frac{(1.054571817 \times 10^{-34} \, \text{Js})^2}{(9.1093837139 \times 10^{-31} \, \text{kg}) (5.2917720989499097 \times 10^{-11} \, \text{m})^2} \\
    &= 1 \cdot E_h \\
    &\Rightarrow \{E_h\}_{HAU} = 1; \qquad [E_h]_{HAU} = 4.3597447222071 \times 10^{-18} \, \text{J} = E_h
\end{align*}

\begin{align*}
E_h &= \frac{\{\hbar\}_{RAU}^2}{\{m_e\}_{RAU} \{a_0\}_{RAU}^2} \; \frac{ [\hbar]_{RAU}^2}{ [m_e]_{RAU} [a_0]_{RAU}^2} \\
    &= \frac{X_h^2}{X_m \cdot X_a^2} \, \frac{( r_h \, 1.054571817 \times 10^{-34} \, \text{Js})^2}{( r_m \, 9.1093837139 \times 10^{-31} \, \text{kg}) ( r_a \, 5.2917720989499097 \times 10^{-11} \, \text{m})^2} \\
    &= \frac{X_h^2}{X_m \cdot X_a^2} \cdot \frac{r_h^2}{r_m \, r_a^2} \cdot E_h \\
    &\Rightarrow \{E_h\}_{RAU} = \frac{X_h^2}{X_m \cdot X_a^2}; \qquad [E_h]_{RAU} = \frac{r_h^2}{r_m \, r_a^2} \cdot E_h = r_{en} \cdot E_h
\end{align*}

### Derived units

- Velocity:
  Atomic unit of velocity $v_0$

\begin{align*}
    v_0 &= \frac{a_0 E_h}{\hbar} \\
        &= \frac{\{
E_h\}_{SI} \{a_0\}_{SI}}{\{\hbar\}_{SI}} \frac{[E_h]_{SI} [a_0]_{SI}}{ [\hbar]_{SI}} \\
        &= \frac{(4.3597447222071 \times 10^{-18}) (5.2917720989499097 \times 10^{-11})}{(1.054571817 \times 10^{-34})} 
        \frac{\text{J} \cdot \text{m}}{\text{Js}} \\
        &= 2.18769126364 \times 10^6 \, \text{m/s} \\
        &\Rightarrow \{v_0\}_{SI} = 2.18769126364 \times 10^6; \qquad [v_0]_{SI} = \text{m/s}
\end{align*}

\begin{align*}
    v_0 &= \frac{\{E_h\}_{HAU} \{a_0\}_{HAU}}{\{\hbar\}_{HAU}} \frac{ [E_h]_{HAU} [a_0]_{HAU}}{ [\hbar]_{HAU}} \\
        &= \frac{1 \cdot 1}{1} \frac{(4.3597447222071 \times 10^{-18} \text{J}) (5.2917720989499097 \times 10^{-11} \text{m})}{(1.054571817 \times 10^{-34} \text{Js})} \\
        &= 1 \cdot v_0 \\
        &\Rightarrow \{v_0\}_{HAU} = 1; \qquad [v_0]_{HAU} = 2.18769126364 \times 10^6 \, \text{m/s} = v_0
\end{align*}

\begin{align*}
    v_0 &= \frac{\{E_h\}_{RAU} \{a_0\}_{RAU}}{\{\hbar\}_{RAU}} \frac{ [E_h]_{RAU} [a_0]_{RAU}}{ [\hbar]_{RAU}} \\
        &= \frac{X_{en} X_a}{X_\hbar} \cdot \frac{r_{en} r_a}{r_h} \cdot v_0 \\
        &\Rightarrow \{v_0\}_{RAU} = \frac{X_{en} X_a}{X_\hbar}; \qquad [v_0]_{RAU} = \frac{r_{en} r_a}{r_{h}} \cdot v_0 = r_{vel} \cdot v_0
\end{align*}

# Set unified grid

```python
import numpy as np

uxgrid_num = 4*1024
uxgrid_max = 50
uxgrid_min = -50

uxgrid_hus, uxgrid_hus_dx = np.linspace(uxgrid_min*(hus.ureg("nm")/hus.length_unit.to("nm")).magnitude,\
                          uxgrid_max*(hus.ureg("nm")/hus.length_unit.to("nm")).magnitude,\
                          uxgrid_num,retstep=True)

uxgrid_hus_width = (uxgrid_hus[-1] - uxgrid_hus[0])
print(uxgrid_hus_dx,uxgrid_hus_width,uxgrid_hus_width*hus.length_unit.to('nm'))
uxgrid_hus_step_size = uxgrid_hus_dx*hus.length_unit.to("nm")
print(uxgrid_hus_step_size)

uxgrid_rus, uxgrid_rus_dx = np.linspace(uxgrid_min*(rus.ureg("nm")/rus.length_unit.to("nm")).magnitude,\
                          uxgrid_max*(rus.ureg("nm")/rus.length_unit.to("nm")).magnitude,\
                          uxgrid_num,retstep=True)

uxgrid_rus_width = (uxgrid_rus[-1] - uxgrid_rus[0])
print(uxgrid_rus_dx,uxgrid_rus_width,uxgrid_rus_width*rus.length_unit.to('nm'))
uxgrid_rus_step_size = uxgrid_rus_dx*rus.length_unit.to("nm")
print(uxgrid_rus_step_size)
```

# Set model-potential

```python
import matplotlib.pyplot as plt
import potentials as pots
import pint


well_depth = 1.0 # eV
well_depth_hus = (well_depth*hus.ureg("eV")/hus.energy_unit.to("eV")).magnitude
well_depth_rus = (well_depth*rus.ureg("eV")/rus.energy_unit.to("eV")).magnitude

well_width = 20.0 # nm
well_width_hus = (well_width*hus.ureg("nm")/hus.length_unit.to("nm")).magnitude
well_width_rus = (well_width*rus.ureg("nm")/rus.length_unit.to("nm")).magnitude


modelpot_hus=pots.SmoothStackPotential(uxgrid_hus,wallwidth=0,\
                                   wallheight=0,\
                                   welldepth=well_depth_hus,\
                                   wellwidth=well_width_hus,\
                                   wallrise = 2,\
                                   wellfall = 2)

modelpot_rus=pots.SmoothStackPotential(uxgrid_rus,wallwidth=0,\
                                   wallheight=0,\
                                   welldepth=well_depth_rus,\
                                   wellwidth=well_width_rus,\
                                   wallrise = 2,\
                                   wellfall = 2)

plt.plot(uxgrid_hus,modelpot_hus(0))
plt.plot(uxgrid_rus,modelpot_rus(0))
plt.xlabel("x [arb. u.]")
plt.ylabel("Pot. energy [arb. u.]")
plt.grid()
plt.show()
```

```python
x = uxgrid_hus * hus.length_unit.to('nm')
y = modelpot_hus(0) * hus.energy_unit.to('eV')
y2 = modelpot_rus(0) * rus.energy_unit.to('eV')

plt.plot(x, y)
plt.plot(x, y2)
plt.xlabel("x [nm]")
plt.ylabel("Pot. energy [eV]")
plt.grid()
plt.show()
```

# Compute eigen states

```python
import potentials as pots
import quantum_toolkit as quat
import scipy.sparse as sparse

zeropot_hus = pots.ZeroPotential(uxgrid_hus)
zeropot_rus = pots.ZeroPotential(uxgrid_rus)

ham_hus = quat.HamiltonOperator(uxgrid_hus,hbar=hus.hbar,\
                                   me=hus.m,\
                                   charge=hus.e,\
                                   scalarpot=modelpot_hus, vectorpot=zeropot_hus )

ham_rus = quat.HamiltonOperator(uxgrid_rus,hbar=rus.hbar,\
                                   me=rus.m,\
                                   charge=rus.e,\
                                   scalarpot=modelpot_rus, vectorpot=zeropot_rus )

eigvals_hus, eigvecs_hus = sparse.linalg.eigsh(ham_hus(0), 20,which = 'SR')
eigvals_rus, eigvecs_rus = sparse.linalg.eigsh(ham_rus(0), 20,which = 'SR')
```

```python
print(ham_hus.diags[1])
print(ham_rus.diags[1])

plt.plot(np.arange(len(ham_hus.diags[1])),ham_hus.diags[1])
plt.plot(np.arange(len(ham_rus.diags[1])),ham_rus.diags[1])
plt.xlabel("n")
plt.ylabel("E")
plt.grid()
plt.show()
```

```python
plt.scatter(np.arange(20),eigvals_hus)
plt.scatter(np.arange(20),eigvals_rus)
plt.xlabel("n")
plt.ylabel("E")
plt.grid()
plt.show()
```

```python
plt.scatter(np.arange(20),eigvals_hus*hus.energy_unit.to('eV'))
plt.scatter(np.arange(20),eigvals_rus*rus.energy_unit.to('eV'))
plt.xlabel("n")
plt.ylabel("E [eV]")
plt.grid()
plt.show()
```

# Check dispersion

```python
disp_hus=quat.Dispersion(uxgrid_hus_dx,hbar=hus.hbar, mass=hus.m)

en_max = 6 #eV
ukgrid_num = 41

en_max_hus = (en_max*hus.ureg("eV")/hus.energy_unit.to("eV")).magnitude
k_max_hus = disp_hus.energy2k(en_max_hus)

ukgrid_hus_max = k_max_hus
ukgrid_hus_min = -k_max_hus

print(f'k_max_hus {k_max_hus}')

ukgrid_hus, ukgrid_hus_dx = np.linspace(ukgrid_hus_min, ukgrid_hus_max,\
                               ukgrid_num,retstep=True)

ukgrid_hus_width = (ukgrid_hus[-1] - ukgrid_hus[0])
print(ukgrid_hus_dx,ukgrid_hus_width,ukgrid_hus_width*hus.wavenumber_unit.to('1/nm'))


disp_rus=quat.Dispersion(uxgrid_rus_dx,hbar=rus.hbar, mass=rus.m)

en_max_rus = (en_max*rus.ureg("eV")/rus.energy_unit.to("eV")).magnitude
k_max_rus = disp_rus.energy2k(en_max_rus)

ukgrid_rus_max = k_max_rus
ukgrid_rus_min = -k_max_rus

print(f'k_max_rus {k_max_rus}')

ukgrid_rus, ukgrid_rus_dx = np.linspace(ukgrid_rus_min, ukgrid_rus_max,\
                               ukgrid_num,retstep=True)

ukgrid_rus_width = (ukgrid_rus[-1] - ukgrid_rus[0])
print(ukgrid_rus_dx,ukgrid_rus_width,ukgrid_rus_width*rus.wavenumber_unit.to('1/nm'))
```

```python
plt.scatter(ukgrid_hus,disp_hus.k2energy(ukgrid_hus))
plt.scatter(ukgrid_rus,disp_rus.k2energy(ukgrid_rus))
plt.xlabel("k")
plt.ylabel("E")
plt.grid()
plt.show()
```

```python
plt.scatter(ukgrid_hus*hus.wavenumber_unit.to('1/nm'),disp_hus.k2energy(ukgrid_hus)*hus.energy_unit.to('eV'))
plt.scatter(ukgrid_rus*rus.wavenumber_unit.to('1/nm'),disp_rus.k2energy(ukgrid_rus)*rus.energy_unit.to('eV'))
plt.xlabel("k [1/nm]")
plt.ylabel("E [h.a.u]")
plt.grid()
plt.show()
```

# Check scattered states

```python
import quantum_transport_toolkit as trat

particle_energy = 0.5 # eV
pe_hus = (particle_energy*hus.ureg("eV") / hus.energy_unit.to("eV")).magnitude
pe_rus = (particle_energy*rus.ureg("eV") / rus.energy_unit.to("eV")).magnitude


pk_hus = disp_hus.energy2k(pe_hus)
pk_rus = disp_rus.energy2k(pe_rus)

pk_hus_scaled = pk_hus*hus.wavenumber_unit.to('1/nm')
pk_rus_scaled = pk_rus*rus.wavenumber_unit.to('1/nm')

print(f'pk_hus {pk_hus} = {pk_hus_scaled}')
print(f'pk_rus {pk_rus} = {pk_rus_scaled}')

qp_pk_hus, qp_mk_hus = trat.quasiparticles_at_given_energy(pe_hus,pk_hus,ham_hus,normed=False)
qp_pk_rus, qp_mk_rus = trat.quasiparticles_at_given_energy(pe_rus,pk_rus,ham_rus,normed=False)
```
